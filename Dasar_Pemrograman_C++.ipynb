```python
%%writefile kelilingsegitiga.cpp
#include <iostream>
using namespace std;
int main()
 {
  float sisiA, sisiB, sisiC, keliling;

  cout << "=== Program Menghitung Keliling segitiga ===" << endl;
  cout << "Masukkan panjang sisi A:  ";
  cin >> sisiA;
  cout << "Masukkan panjang sisi B:  ";
  cin >> sisiB;
  cout << "Masukkan panjang sisi C:  ";
  cin >> sisiC;

  keliling = sisiA  +  sisiB  +  sisiC;

  cout << "Keliling segitiga adalah:   "  << keliling << endl;

  return 0;
}
```
